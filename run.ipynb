```python
import random

def Monty_hall(switch_doors):
    doors = ['Car', 'Goat', 'Goat']
    random.shuffle(doors)
    
    initial_choice = random.choice(range(len(doors)))
    
    if switch_doors:
        doors_revealed = [i for i in range(3) if i !=initial_choice and doors[i] == 'Goat']
        door_revealed = random.choice(doors_revealed)
        
        final_choice = [i for i in range(3) if i != initial_choice and i !=door_revealed][0]
    else:
        final_choice = initial_choice
        
    return doors[final_choice] == 'Car'
```

```python
def simulate_game(num_games):
    num_games_without_switching = sum([Monty_hall(False) for  _ in range(num_games)])
    num_games_with_switching = sum([Monty_hall(True) for  _ in range(num_games)])
    
    return num_games_without_switching/num_games, num_games_with_switching/num_games
```

```python
simulate_game(10)
```
